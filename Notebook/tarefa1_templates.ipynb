```python
import docx
import os

dados_renovacao = {
    # --- Dados Gerais (Partilhados por vários ficheiros) ---
    "{{nome_docente}}": "Professor João Silva",
    "{{periodo_contrato}}": "01/09/2025 a 31/08/2026",
    "{{ano_letivo}}": "2025/2026",

    # --- Ficheiro 1 (Necessidade da Área) ---
    "{{area_contratacao}}": "Engenharia Informática",
    "{{inicio_contrato}}": "01/09/2025",

    # --- Ficheiro 2 (Proposta de Contratação) ---
    "{{funcoes_externas}}": "Não exerce funções externas",

    # --- Ficheiro 3 (Relatório do Júri) ---
    "{{ano_anterior}}": "2024/2025",
    "{{areas_curriculares}}": "Matemática, Física",

    # --- Ficheiro 4 (Ficha de Serviço - Tabela) ---
    # Linha 1 da Tabela
    "{{uc_1}}": "Sistemas Operativos",
    "{{ciclo_1}}": "Licenciatura",
    "{{ano_1}}": "2º",
    "{{sem_1}}": "1º",
    "{{horas_1}}": "6",
    
    # Linha 2 da Tabela
    "{{uc_2}}": "Inteligência Artificial",
    "{{ciclo_2}}": "Licenciatura",
    "{{ano_2}}": "3º",
    "{{sem_2}}": "2º",
    "{{horas_2}}": "6",
    
    # Linha 3 da Tabela (Em branco, para limpar as tags que sobram no Word)
    "{{uc_3}}": "",
    "{{ciclo_3}}": "",
    "{{ano_3}}": "",
    "{{sem_3}}": "",
    "{{horas_3}}": "",
    
    # Total
    "{{total_horas}}": "12"
}

ficheiros_template = {
    "1. Necessidade da área.docx",
    "2. Proposta de contratação final.docx",
    "3.  Relatório_Juri.docx",
    "Ficha de serviço atribuido.docx"
}

pasta_modelos = "../Modelos Contratuais/tempo integral anual"
pasta_output = "../Modelos Contratuais/Modelos Gerados"

#segurança para caso não exista a pasta de output, por qualquer motivo
if not os.path.exists(pasta_output):
    os.makedirs(pasta_output)

def processar_renovacao(lista_ficheiros, dados):
    for nome_ficheiro in lista_ficheiros:
        caminho_template = os.path.join(pasta_modelos, nome_ficheiro)

        if not os.path.exists(caminho_template):
            print(f"Aviso: O ficheiro '{nome_ficheiro}' não foi encontrado na pasta '{pasta_modelos}'.")
            continue

        documento = docx.Document(caminho_template)

        for paragrafo in documento.paragraphs:
            for chave, valor in dados.items():
                if chave in paragrafo.text:
                    paragrafo.text = paragrafo.text.replace(chave, valor)

        #subsituir nas tabelas, crucial para o ultimo ficheiro
        for tabela in documento.tables:
            for linha in tabela.rows:
                for celula in linha.cells:
                    for paragrafo in celula.paragraphs:
                        for chave, valor in dados.items():
                            if chave in paragrafo.text:
                                paragrafo.text = paragrafo.text.replace(chave, valor)

        caminho_output = os.path.join(pasta_output, nome_ficheiro)
        documento.save(caminho_output)
        print(f"Ficheiro '{nome_ficheiro}' processado e salvo em '{caminho_output}'.")

processar_renovacao(ficheiros_template, dados_renovacao)
```
